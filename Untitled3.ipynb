# Chi Sqr test

```python
import numpy as np
```

```python
row1 = np.array([40,45,25,10])
row2 = np.array([35,30,20,30])
```

```python
sum_r1 = np.sum(row1)
sum_r2 = np.sum(row2)
```

```python
sum_row = np.array([sum_r1, sum_r2])
sum_row
```

```python
sum_cal = row1+row2
sum_cal
```

```python
exp = []
for i in sum_row :
  for j in sum_cal:
    value = (i*j)/235
    exp.append(value)
```

```python
exp
```

```python
obj = np.array([40,45,25,10,35,30,20,30])
obj
```

# chi calculated

```python
np.sum(np.square(obj-exp)/exp)
```

```python
dataa = [["s", "m","l","xl"]]
```

## **kind** parameter
1.   quicksort (Default)
2.   mergesort
3.   heapsort






```python
arr = np.array([[7,3,8,6,4], [7,2,9,8,6], [5,4,2,3,1]])

print(np.sort(arr, axis=0, kind='mergesort'))
print()
print(np.sort(arr, axis=1, kind='quicksort'))
print()
print(np.sort(arr, kind='heapsort'))
```
